```python
import matplotlib
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
matplotlib.rcParams['font.family'] = 'AppleGothic'
matplotlib.rcParams['axes.unicode_minus'] = False

fig, ax = plt.subplots(figsize=(14, 6))

milestones = [
    # (month, week_start, week_end, label, color)
    (1, 0, 2, "데이터 파이프라인", "#4472C4"),
    (1, 2, 4, "AutoEncoder 모델", "#4472C4"),
    (2, 4, 6, "signal_tools.py", "#ED7D31"),
    (2, 5, 7, "ReAct 에이전트", "#ED7D31"),
    (2, 7, 8, "Streamlit 대시보드", "#ED7D31"),
    (3, 8, 10, "현장 A/B 테스트", "#A9D18E"),
    (3, 10, 12, "성과 측정 & 보고서", "#A9D18E"),
]

colors_month = {1: "#4472C4", 2: "#ED7D31", 3: "#A9D18E"}

for i, (month, ws, we, label, color) in enumerate(milestones):
    y = len(milestones) - i
    ax.barh(y, we - ws, left=ws, height=0.6, color=color, alpha=0.8)
    ax.text(ws + (we - ws) / 2, y, label, ha='center', va='center', fontsize=9, color='white', fontweight='bold')

ax.set_xlim(0, 12)
ax.set_ylim(0, len(milestones) + 1)
ax.set_xticks([0, 2, 4, 6, 8, 10, 12])
ax.set_xticklabels(['M1 시작', '2주', '4주\nM2', '6주', '8주\nM3', '10주', '12주'])
ax.set_yticks([])
ax.set_title('KAMP CNC 파일럿 3개월 로드맵', fontsize=14, pad=15)
ax.axvline(4, color='gray', linestyle='--', alpha=0.5)
ax.axvline(8, color='gray', linestyle='--', alpha=0.5)

patches = [mpatches.Patch(color=v, label=f"Month {k}") for k, v in colors_month.items()]
ax.legend(handles=patches, loc='lower right')
ax.grid(axis='x', alpha=0.3)

plt.tight_layout()
plt.savefig('kamp_cnc_pilot_roadmap.png', dpi=150, bbox_inches='tight')
plt.show()
print("✅ 로드맵 시각화 완료")
```

```python
def export_pilot_design_md(design: dict) -> str:
    """파일럿 설계서를 마크다운으로 출력"""
    md = f"""# {design['제목']}
**버전**: {design['버전']} | **작성일**: {design['작성일']} | **작성팀**: {design['작성팀']}

---

## 1. 문제 정의
| 항목 | 내용 |
|------|------|
| 현장 문제 | {design['1_문제정의']['현장문제']} |
| 근본 원인 | {design['1_문제정의']['근본원인']} |
| AI 과제 | {design['1_문제정의']['AI_과제']} |
| 기대 효과 | {design['1_문제정의']['기대효과']} |

## 2. 기술 스택
| 영역 | 선택 기술 |
|------|-----------|
"""
    for k, v in design['3_기술스택'].items():
        md += f"| {k} | {v} |\n"
    
    md += "\n## 3. 3개월 로드맵\n"
    for month, details in design['4_3개월로드맵'].items():
        md += f"\n### {details['기간']} — {details['목표']}\n"
        md += f"**성공 기준**: {details['성공기준']}\n"
        md += "**산출물**:\n"
        for item in details['산출물']:
            md += f"- {item}\n"
    
    return md

md_output = export_pilot_design_md(pilot_design)
print(md_output[:1000], "\n\n[... 계속 ...]")
```
